```python
import geopandas as gpd
from bivario import explore_bivariate_data

vn_pop = "https://raw.githubusercontent.com/opengeoshub/vgriddemo/refs/heads/main/pop/vn_pop_h3_6.geojson"

gdf = gpd.read_file(vn_pop)
gdf.head()
```

```python
map = explore_bivariate_data(
    gdf,
    # gdf.to_crs('ESRI:53009'),
    column_a="population",
    column_b="hospital",
    scheme=("NaturalBreaks", "NaturalBreaks"),
    k=(4, 4),
    legend_loc="tr",
    legend_size_px=200,
    column_a_label="Population",
    column_b_label="Hospital",
    zoom_start=5,
    dark_mode=True,   
    cmap="coral_ocean",
    # crs='ESRI53009'
)

map
```
